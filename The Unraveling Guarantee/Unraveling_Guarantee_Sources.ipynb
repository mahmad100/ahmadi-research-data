### Caveats on Data Construction

A few series are not available as a single FRED pull and were assembled
from public Treasury and Federal Reserve sources:

- The 10-year Treasury yield series in Figure 4.1 is built from a
  representative monthly anchor set including the August 2020 0.65% issuance
  cited in the deck, with linear interpolation between anchors.
- Holder composition in Figure 4.3 right panel uses U.S. Treasury TIC
  reports and the Federal Reserve H.4.1 SOMA holdings line.
- Quarterly SOMA unrealized losses and the deferred asset (Figures 3.2 and
  3.3) are hardcoded from the Fed's *Combined Quarterly Financial
  Statements* 2022-2025 because the underlying line items do not have a
  stable FRED series ID.
- CBO scenario lines in Figure 1.2 are cubic-spline interpolations between
  the published anchor years 2024, 2025, 2028, 2035, 2040, 2045, 2050, and
  2055.
