# Palantir Is Down: What I Learned From a First Mini Research — Companion Charts

Reproducible code behind the Ahmadi Research blog post **[Palantir Is Down: What I Learned From a First Mini Research](https://ahmadiresearch.com/blog/palantir-scorecard/)**.

It regenerates the data charts in the post from publicly available sources, with the data inline and the full numbered source list at the bottom. The post revisits a September 2025 short call on Palantir (PLTR) priced at $171.43.

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker

# Brand palette - matches assets/css/style.css on the live site.
BLUE_MAIN  = '#005a96'
BLUE_SOFT  = '#0f80c4'
BLUE_LIGHT = '#8fd1ff'
TITLE_BLUE = '#13597B'
GREY_TEXT  = '#374151'
GREY_GRID  = '#e5e7eb'

plt.rcParams.update({
    'font.family': 'DejaVu Sans',
    'font.size': 10,
    'axes.spines.top': False,
    'axes.spines.right': False,
    'axes.edgecolor': '#9ca3af',
    'axes.labelcolor': GREY_TEXT,
    'xtick.color': GREY_TEXT,
    'ytick.color': GREY_TEXT,
    'axes.grid': True,
    'grid.color': GREY_GRID,
    'grid.linewidth': 0.6,
    'figure.dpi': 110,
    'savefig.dpi': 180,
})
```

## Chart 1 - Price since the call

Palantir's share price from the September 2025 call through July 2026, with the $171.43 short-call level drawn as a dashed reference line. The call point, the November 3, 2025 record close near $207, the June 25, 2026 intraday low near $107, and the July 8, 2026 close of $132.22 are marked. A short entered at $171 would have been underwater for months as the stock first climbed to its peak before falling below entry.

Levels are approximate month-end closes; the annotated call, peak, low, and latest points use the exact figures. Source: Nasdaq; Macrotrends [2].

```python
months = ['Sep 25','Oct 25','Nov 25','Dec 25','Jan 26','Feb 26',
          'Mar 26','Apr 26','May 26','Jun 26','Jul 26']
# Approximate month-end closes. Nov reflects the record close near $207 (Nov 3, 2025);
# Jun reflects the ~$116 month-end after an intraday low near $107 on Jun 25, 2026;
# Jul reflects the $132.22 close on Jul 8, 2026.
price = [171, 188, 202, 178, 172, 165, 160, 158, 157, 116, 132]
CALL = 171.43
key = {0: 'call', 2: 'peak', 9: 'low', 10: 'now'}

fig, ax = plt.subplots(figsize=(10, 4.8))
ax.plot(months, price, color=BLUE_MAIN, linewidth=2.5, zorder=3)
ax.fill_between(months, price, min(price) - 10, color=BLUE_MAIN, alpha=0.10, zorder=1)
ax.axhline(CALL, color=TITLE_BLUE, linewidth=1.4, linestyle='--', zorder=2)
ax.text(len(months) - 1, CALL + 3, 'Short call: $171.43', ha='right', va='bottom',
        fontsize=9, color=TITLE_BLUE, weight='bold')
for i in key:
    ax.plot([months[i]], [price[i]], 'o', markersize=9, color=TITLE_BLUE, zorder=4)

ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda v, _: f'${v:,.0f}'))
ax.set_ylim(90, 215)
ax.set_ylabel('PLTR share price')
ax.set_title(r'Short-called at \$171, it first rose to \$207 before falling below entry',
             color=TITLE_BLUE, fontsize=13, weight='bold', loc='left', pad=14)
plt.setp(ax.get_xticklabels(), fontsize=8.5)
fig.tight_layout()
plt.show()
```

## Chart 2 - The growth the bear case missed

The bear case rested on decelerating growth: the deck cited revenue growth slowing to about 20 percent. Palantir instead reaccelerated, with first-quarter 2026 revenue up 85 percent year over year. A rich multiple can fall two ways, and the company grew into part of its price rather than only falling toward it.

Source: Beyond the Buzz, 2025 [1]; Palantir Q1 2026 results [4].

```python
labels = ["My deck's premise\n(TTM, Sep 2025)", 'Actual\n(Q1 2026)']
growth = [20, 85]

fig, ax = plt.subplots(figsize=(8, 4.8))
bars = ax.bar(labels, growth, color=[BLUE_LIGHT, BLUE_MAIN],
              edgecolor=[BLUE_SOFT, BLUE_MAIN], linewidth=1.3, width=0.6)
for bar, v in zip(bars, growth):
    ax.text(bar.get_x() + bar.get_width() / 2, v + 1.5, f'+{v}%',
            ha='center', va='bottom', fontsize=12, weight='bold', color=TITLE_BLUE)

ax.yaxis.set_major_formatter(mticker.PercentFormatter(decimals=0))
ax.set_ylim(0, 95)
ax.set_ylabel('Revenue growth, year over year')
ax.set_title('The bear case bet on deceleration; growth reaccelerated instead',
             color=TITLE_BLUE, fontsize=13, weight='bold', loc='left', pad=14)
fig.tight_layout()
plt.show()
```

## Chart 3 - Price-to-sales, then vs. now vs. peers

Price-to-sales at the September 2025 call, today, and the current enterprise-software peer median (Salesforce, ServiceNow, Oracle, SAP, Snowflake, Datadog, CrowdStrike). The multiple roughly halved, from about 118 times sales to 59, yet still sits near seven times the peer group, so the reversion to peer levels the thesis implied has barely begun.

Source: Beyond the Buzz, 2025 [1]; company filings [4]; stockanalysis.com [5].

```python
labels = ['At the call\n(Sep 2025)', 'Today\n(Jul 2026)', 'Peer median\n(Jul 2026)']
ps = [118, 59, 8]

fig, ax = plt.subplots(figsize=(8.5, 4.8))
bars = ax.bar(labels, ps, color=[BLUE_MAIN, BLUE_SOFT, BLUE_LIGHT],
              edgecolor=[BLUE_MAIN, BLUE_SOFT, BLUE_SOFT], linewidth=1.3, width=0.6)
for bar, v in zip(bars, ps):
    ax.text(bar.get_x() + bar.get_width() / 2, v + 2, f'{v}x',
            ha='center', va='bottom', fontsize=12, weight='bold', color=TITLE_BLUE)

ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda v, _: f'{v:.0f}x'))
ax.set_ylim(0, 130)
ax.set_ylabel('Price / sales')
ax.set_title('The multiple halved but still sits about seven times the peer median',
             color=TITLE_BLUE, fontsize=13, weight='bold', loc='left', pad=14)
fig.tight_layout()
plt.show()
```

## Chart 4 - Revenue concentration (Q1 2026)

Palantir's first-quarter 2026 revenue split two ways: by customer type (government vs. commercial) and by geography (US vs. international). About 53 percent is government and about 79 percent is US, so the concentration the original deck flagged still holds. International figures are derived from the reported segments.

Source: Palantir Q1 2026 Business Update [11].

```python
cats = ['By customer', 'By geography']
gov_us   = [52.6, 78.5]   # Government / United States
com_intl = [47.4, 21.5]   # Commercial / International
y = np.arange(len(cats))

fig, ax = plt.subplots(figsize=(9, 3.8))
ax.barh(y, gov_us, color=BLUE_MAIN, edgecolor='white', label='Government / United States')
ax.barh(y, com_intl, left=gov_us, color=BLUE_LIGHT, edgecolor='white', label='Commercial / International')
for i in range(len(cats)):
    ax.text(gov_us[i] / 2, y[i], f'{gov_us[i]:.1f}%', ha='center', va='center',
            color='white', fontsize=11, fontweight='bold')
    ax.text(gov_us[i] + com_intl[i] / 2, y[i], f'{com_intl[i]:.1f}%', ha='center', va='center',
            color=TITLE_BLUE, fontsize=11, fontweight='bold')
ax.set_yticks(y)
ax.set_yticklabels(cats, fontsize=11, fontweight='bold')
ax.invert_yaxis()
ax.set_xlim(0, 100)
ax.xaxis.set_major_formatter(mticker.PercentFormatter(decimals=0))
ax.grid(False)
ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.16), ncol=2, frameon=False, fontsize=10)
ax.set_title('Revenue is still concentrated: about half government, four-fifths US',
             color=TITLE_BLUE, fontsize=13, weight='bold', loc='left', pad=14)
fig.tight_layout()
plt.show()
```

## Chart 5 - Dilution

Diluted weighted-average shares outstanding from fiscal 2021 through the first quarter of 2026. Share count rose from about 1.9 billion to about 2.6 billion, an increase of about a third, driven by stock-based compensation rather than any capital raise.

Source: Palantir SEC filings [12].

```python
years  = ['2021', '2022', '2023', '2024', '2025', 'Q1 2026']
shares = [1.92, 2.06, 2.30, 2.45, 2.57, 2.57]

fig, ax = plt.subplots(figsize=(9, 4.4))
bars = ax.bar(years, shares, color=BLUE_MAIN, edgecolor=BLUE_MAIN, width=0.66)
for bar, v in zip(bars, shares):
    ax.text(bar.get_x() + bar.get_width() / 2, v + 0.03, f'{v:.2f}B',
            ha='center', va='bottom', fontsize=10, weight='bold', color=TITLE_BLUE)
ax.set_ylim(0, 2.9)
ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda v, _: f'{v:.1f}B'))
ax.set_ylabel('Diluted weighted-average shares')
ax.set_title('Share count keeps climbing on stock-based comp, not a capital raise',
             color=TITLE_BLUE, fontsize=13, weight='bold', loc='left', pad=14)
fig.tight_layout()
plt.show()
```

## Chart 6 - The shareholder vote

Norway's sovereign wealth fund backed two 2026 proposals asking Palantir to address potential human-rights harms from its work with the Pentagon, the Israeli military, and US immigration enforcement. Close to 60 percent of outside shareholders voted in favour, but the proposals failed: Palantir's founders hold super-voting shares that outweigh the outside vote.

Source: Financial Times [18].

```python
labels  = ['Outside shareholders']
favour  = [60]
against = [40]

fig, ax = plt.subplots(figsize=(9, 2.8))
ax.barh(labels, favour, color=BLUE_MAIN, edgecolor='white', label='In favour')
ax.barh(labels, against, left=favour, color=BLUE_LIGHT, edgecolor='white', label='Against')
ax.axvline(50, color=TITLE_BLUE, linestyle='--', linewidth=1.4)
ax.set_ylim(-0.6, 0.6)
ax.text(50, 0.48, '50% majority', ha='center', va='bottom',
        fontsize=9, color=TITLE_BLUE, fontweight='bold')
ax.text(favour[0] / 2, 0, f'{favour[0]}%', ha='center', va='center',
        color='white', fontsize=12, fontweight='bold')
ax.text(favour[0] + against[0] / 2, 0, f'{against[0]}%', ha='center', va='center',
        color=TITLE_BLUE, fontsize=12, fontweight='bold')
ax.set_xlim(0, 100)
ax.xaxis.set_major_formatter(mticker.PercentFormatter(decimals=0))
ax.set_yticks([])
ax.grid(False)
ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.15), ncol=2, frameon=False, fontsize=10)
ax.set_title('Outside shareholders backed change; founder super-voting shares overruled them',
             color=TITLE_BLUE, fontsize=12, weight='bold', loc='left', pad=16)
fig.tight_layout()
plt.show()
```
